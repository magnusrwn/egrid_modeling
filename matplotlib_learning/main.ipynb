```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# solve:
S = np.array([
    [2, 4],
    [2, 0],
    [1,-2],
])

# determine if these are linearly dependant or independant... or try to.
```

S (set) is linearly independant when:
- the 0th vector exists
- no vector can be made of a sum of other vectors (they can be multiplies by scalars)
    - I.e. each vector is not redundant in its addition to the sets span...

example of linearly dependancy:
a = [4, 3], b = [1, 1], c = [2, 3]

in this case, the set (a,b,c) is linearly dependant, as: ca + cb = c where c is a scalar multiple

```python
origins = np.zeros_like(S)

fig, ax = plt.subplots()

ax.quiver(
    origins[:,0], origins[:,1], # starting x, y
    S[:,0], S[:, 1], # plting vectr dx, dy
    angles="xy", # ?
    scale_units="xy", #?
    scale=1
)
length = np.floor(np.linalg.norm(S)) + 1
ax.set_aspect("equal")
ax.set_xlim(-length, length)
ax.set_ylim(-length, length)
ax.axhline(0, color="black", linewidth=0.5)
ax.axvline(0, color="black", linewidth=0.5)
ax.grid()

# plt.show()
```
