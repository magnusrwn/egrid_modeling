```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
A = np.array([[1,0,0], [0,1,0], [0,0,1]])

fig = plt.figure()
ax = fig.add_subplot(projection="3d")
ax.quiver(
    0,0,0,
    A[:,0],
    A[:,1],
    A[:,2],
)
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")

limit = np.max(np.abs(A)) * 1.1
# gets the max from the aray, and then gives it 10% margin

ax.set_xlim(limit)
ax.set_ylim(limit)
ax.set_zlim(limit)
ax.set_box_aspect((1,1,1)) # creates ratio of all sidex being the same length


plt.show()
```
